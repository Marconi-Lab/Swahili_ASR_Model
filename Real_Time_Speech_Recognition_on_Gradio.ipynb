Import the libbraries

```python
import torch
import gradio as gr
from transformers import pipeline
```

load the swahili model we pretrained

```python
MODEL_NAME = "AntonyG/fine-tune-wav2vec2-large-xls-r-1b-sw"
```

```python
device = 0 if torch.cuda.is_available() else "cpu"
```

Then we put up a pipeline

```python
pipe = pipeline(
    task="automatic-speech-recognition",
    model=MODEL_NAME,
    chunk_length_s=30,
    device=device,
)
```

```python
def transcribe(microphone, state, task="transcribe"):
    file = microphone

    pipe.model.config.forced_decoder_ids = [[2, transcribe_token_id if task=="transcribe" else translate_token_id]]

    text = pipe(file)["text"]

    return state + "\n" + text, state + "\n" + text
```

```python
mf_transcribe = gr.Interface(
    fn=transcribe,
    inputs=[gr.Audio(source="microphone", type="filepath", optional=True), gr.State(value="")],
    outputs=[gr.Textbox(lines=15), gr.State()],
    layout="horizontal",
    theme="huggingface",
    title="🗣️Swahili Automatic Speech Recognition📄",
    live=True,
   description= (" We pretrained xls-r wav2vec 2.0 300m on common voice swahili set. You can now upload a swahili clip, that is less than 30 seconds and get its transcript") )
```

```python
mf_transcribe.launch(enable_queue=True)
```
